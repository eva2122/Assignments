Dataset 1 — CHIRPS Daily Rainfall (Philippines, 1981–present)

UCSB Climate Hazards Group
Compute extreme rainfall metrics (ex days > 100mm, annual max precipitation)

Dataset 2 — NOAA ERSST v5 Sea Surface Temperature

NOAA NCEI
Extract Niño 3.4 index (or Western Pacific SST) and correlate with rainfall extremes

I will use xarray and scipy to compute annual extreme rainfall indexes from CHIRPS data clipped to the Philippine domain, then 
apply grid-cell-level linear trend analysis to detect intensification signals

I will extract the Niño 3.4 SST index from ERSST and correlate it with 
Philippine rainfall variability to assess ENSO's role in driving extremes

Results will be visualized as spatial maps and time series using cartopy 
and matplotlib
