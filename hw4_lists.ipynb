## Part 3: Functions

1. Write a function to convert temperature from kelvin to celsius

```python
def kelvin_to_celsius(temp_k):
    temp_c = temp_k - 273.15
    return temp_c
print(kelvin_to_celsius(300))
```

2. Write a function to convert temperature to fahrenheit
Include an optional keyword argument to specify whether the input is in celcius or kelvin. Call your previously defined functions if necessary.

```python
def convert_to_fahrenheit(temp, unit="C"):
    if unit.upper() == "K":
        temp_c = kelvin_to_celsius(temp)
    else:
        temp_c = temp
    temp_f = (temp_c * 9/5) + 32
    return temp_f
```

3) Check that the outputs are sensible by trying a few examples

```python
print(f"From Kelvin: {convert_to_fahrenheit(300, unit='K')}")
```

```python
print(f"From Celsius: {convert_to_fahrenheit(50, unit='C')}")
```

4) Now write a function that converts from farenheit and uses a keyword argument to specify whether you want the output in celcius or kelvin

```python
def convert_from_fahrenheit(temp_f, out_unit="C"):
    temp_c = (temp_f - 32) * 5/9
    
    if out_unit.upper() == "K":
        return temp_c + 273.15
    else:
        return temp_c
```

```python
print(f"To Celsius: {convert_from_fahrenheit(50, out_unit='C')}")
```

```python
print(f"To Kelvin: {convert_from_fahrenheit(50, out_unit='K')}")
```

5) Write a function that takes two arguments (feet and inches) and returns height in meters

```python
def height_to_meters(feet, inches):
    total_inches = (feet * 12) + inches
    meters = total_inches * 0.0254
    return meters
```

```python
print(f"5'2\" in meters: {height_to_meters(5, 2):.2f}m")
```

```python
print(f"6'5\" in meters: {height_to_meters(6, 5):.2f}m")
```

6. Write a function takes one argument (height in meters) and returns two arguments (feet and inches)

```python
def meters_to_height(meters):
    total_inches = meters / 0.0254
    feet = int(total_inches // 12)
    inches = total_inches % 12
    
    return feet, inches
```

7) Verify that the “round trip” conversion from and back to meters is consistent

```python
f, i = meters_to_height(1.78)
print(f"1.57m is approximately {f} feet and {i:.1f} inches")
```

```python
f2, i2 = meters_to_height(1.83)
print(f"1.96m is approximately {f2} feet and {i2:.1f} inches")
```
